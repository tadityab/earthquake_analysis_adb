```python
from datetime import datetime , timedelta
from pyspark.sql.functions import explode, col, from_unixtime, current_timestamp, concat, lit
import requests
```

```python
def calculate_date(date_val):
    # If date_val is not provided, set it to yesterday's date
    if not date_val:
        yesterday_date = datetime.today() - timedelta(days=1)
        date_val = yesterday_date.strftime('%Y-%m-%d')
    return date_val
```
